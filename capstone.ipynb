```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# FUNCTIONS & MODULAR PROGRAMMING 
def check_pass_status(score):
    """Simple function to determine if a score passes or fails."""
    if score >= 50:
        return "Pass"
    else:
        return "Fail"

# OBJECT-ORIENTED PROGRAMMING (OOP)
class StudentResult:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        self.status = check_pass_status(score)
        
    def get_summary(self):
        return f"Student: {self.name} | Score: {self.score} | Status: {self.status}"

# DATA STRUCTURES & PANDAS
# Creating a simple dataset of study hours vs exam scores
raw_data = {
    'Student_ID': [1, 2, 3, 4, 5],
    'Hours_Studied': [2, 6, 1, 9, 5],
    'Exam_Score': [40, 85, 25, 95, 75]
}
df = pd.DataFrame(raw_data)
print(df, "\n")

# NUMERICAL COMPUTING WITH NUMPY
# Use NumPy to normalize the study hours
hours_array = df['Hours_Studied'].to_numpy()
df['Normalized_Hours'] = hours_array / np.max(hours_array)

# DATA VISUALIZATION WITH MATPLOTLIB
plt.figure(figsize=(6, 4))
plt.scatter(df['Hours_Studied'], df['Exam_Score'], color='blue', s=80, edgecolors='black')
plt.title('Hours Studied vs Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
# SIMPLE CAPSTONE PREDICTION MODEL
# if studied >= 4 hours, predict they will pass (1)
df['Predicted_Pass'] = df['Hours_Studied'].apply(lambda x: 1 if x >= 4 else 0)

print("--- DataFrame with Predictions ---")
print(df[['Student_ID', 'Hours_Studied', 'Exam_Score', 'Predicted_Pass']], "\n")

# Testing our OOP class with an object example
my_record = StudentResult("Ruvimbo", 88)
print("--- OOP Test Output ---")
print(my_record.get_summary())
```
